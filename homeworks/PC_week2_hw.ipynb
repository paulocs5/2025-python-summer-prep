# Q1: Alphabet Slices
* Store the first ten letters of the alphabet in a list.
* Use a slice to print out the first three letters of the alphabet.
* Use a slice to print out any three letters from the middle of your list.

```python
# Solve Alphabet slices here. 
## Extra Credit:  Do this without 'hard coding' the alpahbet.

import string

# Store the first 10 letters

alphabet = list(string.ascii_lowercase[:10])

print("First 10 characters:", alphabet)
print()

# Print out the first three letters of the alphabet
first_three = alphabet[:3]  
print("First three letters:", first_three)

# Print out any three letters from the middle of your list
middle_three = alphabet[3:6]  
print("Three letters from the middle:", middle_three)

# examples:
print("Extra example slice (indices 4-6):", alphabet[4:7])  
print("Another middle slice (indices 2-4):", alphabet[2:5]) 

```

# Q5: 
## Write a function that takes a list and returns a new list that has all the duplicates removed.

Example input and expected output:
- input = `["Michele", "Robin", "Sara", "Michele"]`
- expected output = `['Michele', 'Robin', 'Sara']`


```python
# Solve problem here:

names = ["Michele", "Robin", "Sara", "Michele"]

def remove_duplicates(names):
    return list(set(names))

unique_name = remove_duplicates(names)
print("Original list:", names)
print("No duplicates names:", unique_name)
```

# Q6: Write a function that takes a list of numbers 
(for example, `a = [5, 10, 15, 20, 25]`) and returns a new list of only the first and last elements of the given list.

Example input and expected output:
- input = `[5, 10, 15, 20, 25]`
- expected output = `[5, 25]`

```python
# Solve problem here:
input_list = [5, 10, 99, 20, 25]

def get_first_and_last_elements(lst):
    return [lst[0], lst[-1]]

result = get_first_and_last_elements(input_list)
print("Original:", input_list)
print("First and last:", result)
```

# Q8: Write a function that takes a number as input and returns the following:
* If the input is divisible by three, return `'fizz'`
* If the input is divisible by five, return `'buzz'`
* If the input is divisible by three and by five, return `'fizzbuzz'`
* If the input is not divisible by three or five, return `None`.

```python
# Solve Problem fizzbuzz here:
def fizz_buzz(num):
    if num % 3 == 0 and num % 5 == 0:
        return 'fizzbuzz'
    elif num % 3 == 0:
        return 'fizz'
    elif num % 5 == 0:
        return 'buzz'
    else:
        return None
    
# Tests
print(fizz_buzz(40))
print(fizz_buzz(55))
print(fizz_buzz(33))
print(fizz_buzz(30))
print(fizz_buzz(3))
print(fizz_buzz(5))
print(fizz_buzz(60))

```
